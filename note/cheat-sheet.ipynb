## 1. Opening a terminal, and knowing where you are

Press ⌘-Space, type `Terminal`, press Return. A window opens with a prompt
waiting for you. Every command on this sheet is typed there, one line at a time,
each followed by Return.

The most common mistake in this course is running a git command in the wrong
folder. git only knows about the folder you are standing in, so before anything
else, find out where that is:

```bash
pwd             # print working directory -- where am I?
ls              # list -- what is in here?
cd ~/phy320     # change directory -- go to the course folder
cd ..           # go up one, into the folder that contains this one
```

`~` is your home folder, so `~/phy320` means `phy320` inside it. If you put the
course somewhere else and cannot remember where, drag the folder out of Finder
and drop it onto the Terminal window: its full path appears at the prompt, and
you can type `cd ` in front of it.

## 2. The rhythm

Six commands, in this order, every week. The comments are what each one actually
does.

```bash
cd ~/phy320
	# Everything below happens inside the course folder.

git pull materials main
	# Collect the notes and assignments I have published since
	# you last looked.  Start here every time.

uv run jupyter lab
	# Open Jupyter, and do the work.
	# Ctrl-C in this terminal when you are finished.

git add assignment/03-odes/odes.ipynb
	# Choose what goes into the savepoint.

git commit -m "pendulum RHS, and one RK4 step"
	# Make the savepoint, on your laptop.

git push
	# Send your savepoints to your own repository.
	# This IS the submission.  There is no other button.
```

Name whichever notebook you are actually working on; `03-odes` above is only an
example.

Three things worth understanding about that list.

**`add` and `commit` are two steps on purpose.** `add` says which changes belong
together and `commit` records them with a message, so you can `add` three files
and `commit` them as a single savepoint.

**Committing is not submitting.** A commit lives on your laptop and nowhere
else. Until you `push`, I cannot see it — and neither can a replacement laptop.
Push early and push often: I grade the last thing you pushed before the
deadline, so anything pushed is safe.

**Commit when a piece works, not once at the end.** Then when the next change
breaks everything, you have a place to retreat to. Write messages that say what
changed, so that `git log` reads as an account of your thinking.

## 3. `git status` — the command to type when you are lost

Type it whenever you are unsure what is going on. It is safe, it changes
nothing, and it tells you all of it at once: which branch you are on, whether
you are ahead of your repository, which files you changed, and which changes are
staged for the next commit.

```bash
git status              # what is going on
git status -s           # the same, one short line per file
git diff                # what you changed and have not yet added.  q to leave
git log --oneline -5    # your last five savepoints, newest first.  q to leave
```

What it is telling you:

| `git status` says | it means |
|---|---|
| `Changes not staged for commit` | You edited these. `git add` the ones you want in the next commit. |
| `Changes to be committed` | Staged and ready. `git commit -m "..."` records them. |
| `Untracked files` | git has never seen these before. `git add` them, or leave them alone. |
| `Your branch is ahead of 'origin/main' by 3 commits` | You committed but never pushed. The work exists only on this laptop. `git push`. |
| `nothing to commit, working tree clean` | Everything is committed. Worth running `git push` anyway, to be sure it is sent. |

## 4. When something goes wrong

Nothing in this table is a disaster and nothing here loses your work. Find your
message in the left column; the right column is what to do about it.

| the message | what to do |
|---|---|
| `fatal: not a git repository` | You are in the wrong folder. `cd ~/phy320` and try again. |
| `Permission denied (publickey)` | GitHub cannot tell who you are. Test it with `ssh -T git@github.com`, then work through [GitHub's troubleshooting page](https://docs.github.com/en/authentication/troubleshooting-ssh/error-permission-denied-publickey). |
| `Updates were rejected because the remote contains work that you do not have locally` | I pushed something to your repository — usually a released solution — since you last pulled. `git pull`, then `git push`. |
| `Your branch is ahead of 'origin/main' by N commits` | Not an error. You committed and did not push. `git push`. |
| `Your local changes to the following files would be overwritten by merge` | You are mid-edit and pulling on top of it. Commit first: `git add -A`, `git commit -m "work in progress"`, then `git pull`. |
| `Please tell me who you are` | git has no name on this machine. Run the two `git config --global user.name` / `user.email` lines from section 1 of `getting-set-up`. |
| `CONFLICT (content): Merge conflict in ...` | Read the next paragraph. |

**A merge conflict in a notebook.** git was handed two versions of the same cell
and would not choose between them for you. It is an ordinary event, not damage,
and nothing has been lost.

```bash
git status              # which files are conflicted
uv run nbdiff           # what actually differs, cell by cell
```

Open the notebook, make each conflicted cell say what you want it to say, then
`git add` that file and `git commit`. If you would rather not deal with it right
now, `git merge --abort` puts everything back exactly as it stood before the
pull, and nothing is lost by doing that.

Bring your first one to office hours regardless. Resolving conflicts is a skill,
not an emergency, and you will meet it again in every job you ever have.

## 5. Looking it up

- **[dangitgit.com](https://dangitgit.com)** — "I did something terrible, how do
  I fix it?", arranged by what you did rather than by what the command is
  called. The fastest route from broken to unbroken.
- **[Pro Git, chapter 2](https://git-scm.com/book/en/v2/Git-Basics-Recording-Changes-to-the-Repository)** —
  the free official book. Chapter 2 is exactly the add / commit / push loop in
  section 2 above, explained properly. Worth reading once this semester.
- **[git-scm.com/docs](https://git-scm.com/docs/git-status)** — the reference
  page for every git command. Dense, and authoritative.
- **[nbdime](https://nbdime.readthedocs.io/en/latest/)** — what makes notebook
  diffs and conflicts readable. `uv run nbdiff-web` opens a side-by-side
  comparison of two notebooks in your browser.

And ask me. A screenshot of the terminal — the command you ran *and* everything
it printed — tells me far more than "git isn't working", and it is the single
most useful thing you can bring to office hours.
