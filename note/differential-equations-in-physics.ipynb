# Newton's Laws

1.  There are inertial reference frames.
2.  $m \vec{a} = \sum \vec{F}$ which can be understood as a differential equation, $m\times d^2\vec{x}/dt^2 = \sum \vec{F}$.
3.  Equal and opposite forces.

When we say we're interested in 'solving' Newton's laws, what we mean is that we would like to know the position $\vec{x}(t)$ as a function of time given the forces involved.  But the forces $F$ can depend on the position $\vec{x}$ and the velocity $v=dx/dt$.

Newton's 2nd law is a differential equation, meaning that it relates derivatives of $x$ to a function of $x$ itself.  Checking a function is a solution to N2 is 'as simple' as differentiating it and plugging it in on both sides, and seeing that the equation is true.

The issue is: how do you even find a function that you want to check?  The answer is usually "experience" (someone else solved found this solution and told us and we learned it).

In fact, MOST scenarios don't have so-called "exact" solutions; an "exact solution" is one where you've got a simple formula for $x(t)$.

Instead of trying to capture all of the complication of the world (or of the complicated behavior of some simple object) in a horrible formula, instead what we can do is *integrate* N2 numerically.

# A First Order Differential Equation (an equation with only 1st derivatives in it)

Some examples:
\begin{align}
    dx/dt = 3 + x + x^2
\end{align}
A *solution* to this equation is a *function* $x(t)$ whose first derivative (LHS) matches $3 + x(t) + x(t)^2$ (RHS).

This equation is *solved* by
\begin{align}
    x(t) &= \frac{1}{2}\left(-1 + \sqrt{11} \tan\left(\frac{1}{2}(\sqrt{11}t + \sqrt{11} c_1)\right)\right)
\end{align}

With that $x(t)$ we can calculate
\begin{align}
    \frac{dx}{dt} &= \frac{\sqrt{11}}{2} \sec^2\left(\frac{1}{2}(\sqrt{11}t + \sqrt{11} c_1)\right) \times \frac{\sqrt{11}}{2}
     = \frac{11}{4} \sec^2\left(\frac{\sqrt{11}}{2}(t + c_1)\right)
     \\
     3 + x + x^2 &= 3 + \frac{1}{2}\left(-1 + \sqrt{11} \tan\left(\frac{1}{2}(\sqrt{11}t + \sqrt{11} c_1)\right)\right) + \left[\frac{1}{2}\left(-1 + \sqrt{11} \tan\left(\frac{1}{2}(\sqrt{11}t + \sqrt{11} c_1)\right)\right)\right]^2
     \nonumber\\
     &= \frac{5}{2} + \frac{1}{2}\sqrt{11} \tan \theta + \frac{1}{4} \left[-1 + \sqrt{11} \tan\theta \right]^2
     \nonumber\\
     &= \frac{5}{2} + \frac{1}{2}\sqrt{11} \tan \theta + \frac{1}{4} \left[1^2 - 2 \sqrt{11} \tan\theta + 11 \tan^2\theta \right]
     \nonumber\\
     &= \frac{5}{2} + \frac{1}{4} \left[1^2 + 11 \tan^2\theta \right]
    \nonumber\\
     &= \frac{11}{4} + \frac{11}{4} \tan^2\theta
         \nonumber\\
     &= \frac{11}{4} (1 + \tan^2\theta) = \frac{11}{4} \sec^2\theta
\end{align}
by trigonometric identity!  So, $x(t)$ really does solve our differential equation: as a function of time (ie. use any $t$ you like) and $c_1$ lets us adjust $x(t=0)$!

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def x(c, t):
    return 1/2 * (-1 + np.sqrt(11) * np.tan( np.sqrt(11)/2 * (t + c)))
```

If someone tells us how the system starts out---in other words, they tell us a value for $x(t=0)$, then we have to figure out which $c$ to use, and once we figure out $c$ that determines the solution completely.

```python
t = np.linspace(-4, 4, 1000)
plt.plot(t, x(0, t))
plt.plot(t, x(0.1, t))
plt.ylim(-5, 5)
```

# A Simple 1st-order equation

A good PHY 241 equation that describes motion with constant acceleration, projectile motion, etc.
We're provided with the velocity as a function of time
\begin{align}
    \frac{dx}{dt} &= v_0 + a t
\end{align}
with $v_0$ and $a$ time-independent constants.

This is solved by the position as a function of time,
\begin{align}
    x(t) &= x_0 + v_0 t + \frac{1}{2} a t^2
\end{align}
for any constant $x_0$.

We can check by differenating
\begin{align}
    \frac{dx}{dt} &= \frac{d x_0}{dt} + \frac{d}{dt}(v_0 t) + \frac{d}{dt}\left(\frac{1}{2} a t^2\right) 
    \nonumber\\
    &= 0 + v_0 + \frac{1}{2} a \times 2t = v_0 + a t
\end{align}

The $x_0$ constant is the starting place for the motion; in other words it's the value of $x(t=0)$!

Aside from 'experience' we can solve this differential equation by integrating both sides with respect to time!
\begin{align}
    \frac{dx}{dt} &= v_0 + a t
    \\
    \int_{x(0)}^{x(t)} dx = \int_0^t \frac{dx}{dt}\; dt &= \int_0^t (v_0 + at)\; dt = \int_0^t v_0\; dt + \int_0^t at\; dt
    \\
    &= v_0\int_0^t 1 \; dt + a\int_0^t t^1\; dt
    \\
    x(t) - x(0) &= v_0\int_0^t t^0\; dt + a\int_0^t t^1\; dt = v_0 \left.t\right|^t_0 + a \left.\left(\frac{1}{2} t^2\right)\right|_0^t
    \\
    x(t) &= x(0) + v_0 t + \frac{1}{2} a t^2!
\end{align}

because by the power rule $\frac{d}{dt} (t^n) =  n t^{n-1}$ leads us to $\frac{1}{2} \frac{d}{dt} t^2 = t^1$ and $\frac{d}{dt} t = 1$.

In this case we can also differentiate and find the acceleration by differentiating both sides,
\begin{align}
    \frac{d}{dt} \frac{dx}{dt} &= \frac{d}{dt}\left(v_0 + a t \right)
    \\
    \frac{d^2x}{dt^2} &= 0 + a = a
\end{align}

This summarizes a huge fraction of PHY 241 in terms of differential equations: if the acceleration is constant, $d^2x/dt^2 = a$, then the velocity is linear, $dx/dt = v_0 + a t$, and the position is quadratic $x(t) = x_0 + v_0 t + \frac{1}{2} a t^2$.

To go from $d^2x/dt^2 = a$ (a constant acceleration) to $dx/dt = v_0 + a t$ we integrated both sides, and the initial condition $v_0$ plays the role of the "+C" when integrating.

Similarly, to go from $dx/dt = v_0 + at$ to $x(t) = x_0 + v_0 t + 1/2 a t^2$ we integrated both sides and the initial condition $x_0$ plays the role of "+C" when integrating.

Or, if you prefer, the initial condition stems from doing definite integration on both sides of the equation---on one side we integrate from the starting *time* to the ending *time* and on the other side we integrate from the starting *place* to the ending *place* (just like we would in a $u$-substitution).

# Recap

The two differential equations we studied (so far)
\begin{align}
    \frac{dx}{dt} &= v_0 + a t
    &
    \frac{dx}{dt} &= 3 + x + x^2
\end{align}
have really different seeming properties.
The first only has a function of t on the RHS while the second has a function of x on the RHS.
This makes the difference when trying to integrate.  We can't integrate the second equation with respect to time so easily because we don't know $x(t)$ that appears on the RHS.

If you take the second equation and divide both sides by the RHS you can try the 'just integrate both sides' method:
\begin{align}
    \int_0^t \frac{1}{3 + x + x^2} \frac{dx}{dt}\; dt &= \int_0^t dt
\end{align}
You *might* be able to find the right u-substitution that transforms the LHS into a doable integral, ~but I myself cannot do that~.  Let's do the u-sub $x(t)$, $dx = dx/dt \times dt$
\begin{align}
    \int_0^t \frac{1}{3 + x + x^2} dx &= \int_0^t dt
\end{align}

It's easy to make up differential equations which cannot be so easily 'separated' and thereby reduced to possibly-difficult but 'elementary' integrals.

# The Geometric Meaning of a 1st Order ODE

What we're given is a function that determines the slope of the function interest.

```python
v_0, a = 1, 2
# exactly the same as
# v_0 = 1
# a = 2
def slope(t, x):
    return v_0 + a * t
```

We'd like to be able to plot $x(t)$ but all we know is $dx/dt = $`slope(t)`.

We can plot the *slope field* which we can compute even without knowing the solution!  We only need the differential equation to plot the slope field!

A first-order ordinary differential equation (ordinary: 1 independent variable) for a function $x(t)$ is any equation of the form
\begin{align}
    \frac{dx}{dt} &= f(x, t)
\end{align}
and this gives us a natural interpretation as a *slope field*.  The *slope* of x is determined by the *value* of x as well as the variable t.

```python
def plot_slope_field(t, x, dx_dt, ax):
    T, X = np.meshgrid(t, x)

    # tan(theta) = slope
    # theta = arctan(slope)
    THETA = np.arctan(dx_dt(T, X))
    U = np.cos(THETA)
    V = np.sin(THETA)

    ax.quiver(T, X, U, V, color="black", angles='xy', pivot='middle')
```

```python
t = np.linspace(-4, 4, 20)
x = np.linspace(-4, 4, 20)

fig, ax = plt.subplots()
plot_slope_field(t, x, slope, ax)

ax.set(xlim=(-5, 5), ylim=(-5, 5))
ax.set_xlabel('t')
ax.set_ylabel('x')
```

The *solutions* to the differential equation are tangent to the slope field at every point.

```python
t = np.linspace(-4, 4, 20)
x = np.linspace(-4, 4, 20)

fig, ax = plt.subplots()
plot_slope_field(t, x, slope, ax)

ax.set(xlim=(-5, 5), ylim=(-5, 5))
ax.set_xlabel('t')
ax.set_ylabel('x')

# Let's plot the solution for many different initial conditions.
# They're also called integral curves.
for x_0 in (-4, -2, 0, 2, 4):
    ax.plot(t, x_0 + v_0 * t + 0.5 * a * t**2, label=x_0)
ax.legend()
```

The solution to a 1st order differential equation is called an *integral curve* and that curve's property is that its slope matches the slope field at every place!

# The Exponential Curve

The *exponential* function $e^{kx}$ satisfies the equation
\begin{align}
    \frac{df}{dx} &= k f(x)
\end{align}

Since
\begin{align}
    e^x = \sum_{n=0}^\infty \frac{x^n}{n!} \approx 1 + x + \frac{x^2}{2} + \frac{x^3}{3!} + \cdots
\end{align}
and we can't add pure numbers, lengths, areas, volumes, and so on, $x$ must be dimensionless.

Let's divide both sides by $f(x)$ and integrate both sides $dx$ from $x=0$ to $x=X$,
\begin{align}
    \int_0^X \frac{1}{f} \frac{df}{dx} \; dx &= \int_0^X k \; dx
\end{align}

Let's observe that right-hand side is very simple---the integral of a constant.
\begin{align}
    \int_0^X k \; dx &= k x\big|_{0}^X = kX - k 0 = k X
\end{align}

The left-hand side is also simple---think of it as a u-substitution.
\begin{align}
    \int_0^X \frac{1}{f} \frac{df}{dx} &= \int_{f(0)}^{f(X)} \frac{df}{f} = \ln f \big|_{f(0)}^{f(X)} = \ln f(X) - \ln f(0) = \ln \frac{f(X)}{f(0)}
\end{align}
because $\log a - \log b = \log (a/b)$

\begin{align} 
    a &= e^{\ln a}
    \\
    b &= e^{\ln b}
    \nonumber\\
    e^{\ln(ab)} = ab &= e^{\ln a} e^{\ln b} = e^{\ln a + \ln b}
    \\
    e^{\ln(a/b)} = \frac{a}{b} &= \frac{e^{\ln a}}{e^{\ln b}} = e^{\ln a - \ln b}
\end{align}
because positive exponents in the denominator are negative exponents in the numerator.

Putting the two sides equal to one another,
\begin{align}
    \ln \frac{f(X)}{f(0)} &= k X
    &
    \frac{f(X)}{f(0)} &= e^{k X}
    &
    f(X) &= f(0) e^{k X}
\end{align}

# The Logistic Curve

The [*logistic function*](https://en.wikipedia.org/wiki/Logistic_function), *logistic curve*, or *sigmoid* obeys a different kind of first-order differential equation,
\begin{align}
    \frac{dx}{dt} &= x(t) \left(1-x(t) \right)
\end{align}

```python
def logistic_slope(t, x):
    return x * (1-x)
```

```python
fig, ax = plt.subplots()
t = np.linspace(-4, 4, 20)
x = np.linspace(-1, 2, 40)
plot_slope_field(t, x, logistic_slope, ax)
ax.set_ylim(-1, 2)

ax.set_xlabel('t')
ax.set_ylabel('x')
```

From the slope field we can see that the constant function $x(t) = 1$ is a solution to the logistic differential equation
\begin{align}
    \frac{dx}{dt} &= x(t) \left(1-x(t) \right)
\end{align}
and the constant function $x(t) = 0$ is a solution too!  We can verify these solutions by evaluating both sides of the differential equation and seeing that they match (in both cases, $0=0$).

Let's rewrite $x(t) = 1-u(t)$,
\begin{align}
    x(t) &= 1-u(t)
    &
    1-x(t) &= 1 - (1-u(t)) = u(t)
    &
    \frac{dx}{dt} &= \frac{d}{dt}(1-u(t)) = - \frac{du}{dt}
\end{align}
so that the differential equation becomes
\begin{align}
    -\frac{du}{dt} &= (1-u(t)) u(t)
\end{align}

```python
def first_solution(t):
    return np.ones_like(t)

def second_solution(t):
    return np.zeros_like(t)

fig, ax = plt.subplots()
t = np.linspace(-4, 4, 20)
x = np.linspace(-1, 2, 40)

ax.plot(t, first_solution(t), zorder=-1)
ax.plot(t, second_solution(t), zorder=-1)
plot_slope_field(t, x, logistic_slope, ax)
ax.set_ylim(-1, 2)

ax.set_xlabel('t')
ax.set_ylabel('x')
```

This differential equation has a (relatively) simple solution,
\begin{align}
    x(t) &= \frac{e^{t}}{e^t+C}
\end{align}

```python
def logistic_function(C, t):
    return np.exp(t) / (np.exp(t) + C)
```

```python
def first_solution(t):
    return np.ones_like(t)

def second_solution(t):
    return np.zeros_like(t)

fig, ax = plt.subplots()
t = np.linspace(-4, 4, 20)
x = np.linspace(-1, 2, 40)

ax.plot(t, first_solution(t), zorder=-1)
ax.plot(t, second_solution(t), zorder=-1)
plot_slope_field(t, x, logistic_slope, ax)
ax.set_ylim(-1, 2)

ax.set_xlabel('t')
ax.set_ylabel('x')

t = np.linspace(-4, 4, 200)
for C in (0, 1, 2, 4, 8):
    ax.plot(t, logistic_function(C, t), label=C)

ax.legend()
```

**Integral curves can never cross**.  The reason is that if there were a place (a single $t$ and $x(t)$) where they crossed then they must have different slopes at that place, *but* we know both solutions satisfy the same differential equation and therefore must have the same slope at that place.

The solution $e^t / (e^t + C)$ has the following nice properties:
1.  In the limit $t\to\infty$ the solution goes to $1$!  One way to see that is to multiply top and bottom by $e^{-t}$ so that solution is $ 1 / (1 + C e^{-t}) \to 1$ as $t\to\infty$.
2.  In the limit $t\to-\infty$ the solution goes to $0$.

The solutions 'in between' the asymptotes have $0 \leq C \leq \infty$.  But what about the negative values of $C$?

```python
def first_solution(t):
    return np.ones_like(t)

def second_solution(t):
    return np.zeros_like(t)

fig, ax = plt.subplots()
t = np.linspace(-4, 4, 20)
x = np.linspace(-1, 2, 40)

ax.plot(t, first_solution(t), zorder=-1)
ax.plot(t, second_solution(t), zorder=-1)
plot_slope_field(t, x, logistic_slope, ax)
ax.set_ylim(-1, 2)

ax.set_xlabel('t')
ax.set_ylabel('x')

t = np.linspace(-4, 4, 200)
for C in (-1, -2, -4, -8):
    low = t[t < np.log(-C)]
    ax.plot(low, logistic_function(C, low), label=C)
    high = t[t > np.log(-C)]
    ax.plot(high, logistic_function(C, high), label=C)

ax.legend()
```

The solution $x(t) = e^{t} / (e^t + C)$ can have a denominator that is zero when $C < 0$.  The denominator is zero when $e^t + C = 0$ or $e^t = - C$ or $t = \ln (-C)$.

# Another Example

Consider the differential equation
\begin{align}
    \frac{dy}{dx} = 1 + x - y(x) = f(x, y)
\end{align}
which we can observe has a simple solution $y(x) = x$ since $dy/dx = 1$ and $1+x-y(x) = 1$.

(**Note**: you cannot separate variables.)

```python
def f(x, y):
    return 1 + x - y

fig, ax = plt.subplots()
x = np.linspace(-2, 2, 20)
y = np.linspace(-2, 2, 40)

plot_slope_field(x, y, f, ax)

ax.plot(x, x+1, color='blue', label='slope = 0') # f = 0
ax.plot(x, x,   color='orange', label='slope = 1') # f = 0
ax.plot(x, x-1, color='blue', label='slope = 2') # f = 2


ax.set_xlim(-2,2)
ax.set_ylim(-2,2)
ax.set_xlabel('x')
ax.set_ylabel('y')
```

# Euler's Method

Let's use the geometric meaning of the differential equation
\begin{align}
    \frac{dy}{dx} = f(y, x)
\end{align}
and the understanding of the solutions being integral curves to *find solutions numerically*.

One way to understand what a Taylor series means is that if you take a very small step the function you're expanding doesn't change that much,
\begin{align}
    y(x + \Delta x) &\approx y(x) + y'(x) \Delta x + \cdots
\end{align}
Ignoring the higher-order terms, this leads us to
\begin{align}
    \Delta y = y(x+\Delta x) - y(x) &\approx f(y, x) \Delta x
\end{align}

Suppose we start with $(x_0, y_0)$.  We can evaluate $f(y_0, x_0)$.  If you pick a small $\Delta x$ you receive a small $\Delta y$.
Call $(x_1 = x_0 + \Delta x, y_1 = y_0 + \Delta y)$.  We're walking along a little line segment whose slope is $f(y_0, x_0)$ and whose run is $\Delta x$ and therefore whose rise $\Delta y = f(y_0, x_0) \Delta x$.

```python
def f(x, y):
    return 1 + x - y

x_0 = -1.1
y_0 = 0.314159

fig, ax = plt.subplots()
plot_slope_field(x_0, y_0, f, ax)

Delta_x = 0.1
Delta_y = f(x_0, y_0) * Delta_x

ax.plot(np.array([x_0, x_0 + Delta_x]),
        np.array([y_0, y_0 + Delta_y]),
       )

x_1 = x_0 + Delta_x
y_1 = y_0 + Delta_y
ax.plot(x_1, y_1, marker='o', label='(x_1,y_1)')
plot_slope_field(x_1, y_1, f, ax)

Delta_y = f(x_1, y_1) * Delta_x
x_2 = x_1 + Delta_x
y_2 = y_1 + Delta_y
ax.plot([x_1,x_2], [y_1, y_2])
ax.plot(x_2, y_2, marker='o', label='(x_2,y_2)')


ax.set_xlabel('x')
ax.set_ylabel('y')
ax.legend()
```

The *idea* of Euler's method is that $dy/dx = f(x, y)$ tells us the slope everywhere and we can approximate $y(x)$ as a sequence of little line segments whose slope matches $f(x, y)$ at the left end of each little line segment.  If you make the line segments really short you hardly make a mistake (the mistakes are like $\Delta x^2$).

```python
def euler_step(f, x, y, Delta_x):
    # Differential equation:
    dy_dx = f(x, y)
    # Approximate the rise from the run and the slope
    Delta_y = dy_dx * Delta_x # There's a mistake here, but it's order Delta_x^2.
    return Delta_y
```

```python
steps = 100000
x = np.zeros(steps)
y = np.zeros(steps)

# Our inital starting point
x[0] = 2
y[0] = -2

# How big a step
Delta_x = 0.001

# Use the line-segment extrapolation method (Euler's method)
for step in range(steps-1):
    x[step+1] = x[step] + Delta_x
    Delta_y = euler_step(f, x[step], y[step], Delta_x)
    y[step+1] = y[step] + Delta_y

fig, ax = plt.subplots()
X = np.linspace(-2, 10, 20)
Y = np.linspace(-2, 10, 40)
plot_slope_field(X, Y, f, ax)

ax.plot(x, y)

ax.set_xlim(-2,10)
ax.set_ylim(-2,10)
ax.set_xlabel('x')
ax.set_ylabel('y')
```

The resulting curve (as long as $\Delta x$ is small) matches the slope very well everywhere.  And therefore it's an integral curve (or it's approximately an integral curve) and therefore it's a solution to the differential equation!  (Or it's nearly a solution, admitting we made mistakes like $\Delta x^2$.)

Notice what we did NOT do while building this solution: we did NOT find $y(x) = \text{some function of x}$.  We didn't learn a formula for $y$.  Instead we learned an *algorithm* for constructing $y(x)$.
