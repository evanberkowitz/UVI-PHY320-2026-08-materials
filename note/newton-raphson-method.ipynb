# Newton' Method

[*Newton's Method*](https://en.wikipedia.org/wiki/Newton%27s_method), or the *Newton-Raphson method* is a way to find roots or zeros of functions numerically.

As an example, let's take the parabola

\begin{align}
f(x) &= a x^2 + bx + c
&
\frac{df}{dx} &= 2 a x + b.
\end{align}

whose roots we *could* find with the quadratic formula if we so desired.

```python
# Some choices for the coefficients
a=1
b=-2
c=-1

# And the implementations of f and df/dx
def f(x):
    return a * x**2 + b * x + c

def df(x):
    return 2 * a * x + b
```

The idea of Newton's method is to make a guess that you know isn't right (unless you get wildly lucky) and then to fix it up.

How should we fix it up?  Well, $f$ might be a complicated function, but let's pretend it's a line.  Let's use the slope at our guess to try and fix the guess up!

How should we do that?

Suppose someone hands us a point $p_0=(x_0, y_0)$ and a slope $s_0$.  Then the line that goes through $p_0$ with slope $s_0$ is given by
\begin{align}
    y = y_0 + s_0 (x-x_0).
\end{align}
(This is the 'point-slope' formula; you can check it's right by plugging in $x_0$ and checking that it really produces $y=y_0$ and observing that $dy/dx = s$.

```python
def line(point, slope, x):
    x0 = point[0]
    y0 = point[1]
    return y0 + slope * (x-x0)
```

The function $f$ isn't a line, but if it were we could solve this equation with $y=0$ to find the root $x$,
\begin{align}
    0 &= y_0 + s_0 (x-x_0)
    \\
    -s_0 (x-x_0) &= y_0
    \\
    x-x_0 &= - \frac{y_0}{s_0}
    \\
    x &= x_0 - \frac{y_0}{s_0}.
\end{align}
Without knowing much about the function $f$, this is what we'll guess next.

```python
def next_guess(starting_x, starting_y, starting_slope):
    return starting_x - starting_y / starting_slope
```

The next guess probably isn't a root either!  Let's check:

```python
x = np.linspace(-5,5,1000)
plt.plot(x, f(x), color='black')

g0 = -4

y0 = f(g0)
p0 = (g0, y0)
s0 = df(g0)

plt.plot(g0, y0, color='red', marker='o')
plt.plot(x, line(p0, s0, x), color='red')

g1 = next_guess(g0, y0, s0)
plt.plot(g1, 0, color='orange', marker='o')

plt.ylim((-5,25))
plt.grid()
```

No, no luck.  If we plug our next guess in we can see that it's not a root: the function isn't zero at that $x$ value.

But look: we're closer to a root than we were before.  So let's keep going!

Use our new, next guess as the starting point, find the value of the function and its slope there.

That's the philosophy of Newton's method: we can improve our guess by using the point-slope formula to go to where the root 'ought to be' were the function a line, even though we know it isn't a line!

Let's try iterating a few times by hand.

```python
x = np.linspace(-5,5,1000)
plt.plot(x, f(x), color='black')

g0 = -4

y0 = f(g0)
p0 = (g0, y0)
s0 = df(g0)
plt.plot(g0, y0, color='red', marker='o')
plt.plot(x, line(p0, s0, x), color='red')
g1 = next_guess(g0, y0, s0)
plt.plot(g1, 0, color='red', marker='o')

y1 = f(g1)
p1 = (g1, y1)
s1 = df(g1)
plt.plot(g1, y1, color='orange', marker='o')
plt.plot(x, line(p1, s1, x), color='orange')
g2 = next_guess(g1, y1, s1)
plt.plot(g2, 0, color='orange', marker='o')

y2 = f(g2)
p2 = (g2, y2)
s2 = df(g2)
plt.plot(g2, y2, color='green', marker='o')
plt.plot(x, line(p2, s2, x), color='green')
g3 = next_guess(g2, y2, s2)
plt.plot(g3, 0, color='green', marker='o')

plt.grid()
plt.ylim((-5,25))
```

Use the Taylor series
\begin{align}
    f(x_0  + \Delta x) &\approx f(x_0) + f'(x_0) \Delta x + \frac{1}{2!} f''(x_0) \Delta x^2 + \cdots
\end{align}
We are interested in the value of $\Delta x_0$ such that $f(x_0+\Delta x) = 0$---in other words $x_0+\Delta x$ is a root of $f$!

Ignoring higher-order terms, the Taylor series approximates
\begin{align}
    \Delta x = - \frac{f(x_0)}{f'(x_0)}
\end{align}
Solve for $\Delta x$


```python
def newton_raphson_step(f, df, starting_guess):
    y = f(starting_guess)
    s = df(starting_guess)
    change = - y / s
    return change
```

```python
g = -4

for iteration in range(10):
    print(g, f(g))
    change = newton_raphson_step(f, df, g)
    print('   ',change)
    g += change

print(g, f(g))
```

```python
x = np.linspace(-5,5,1000)
plt.plot(x, f(x), color='black')

g = -4

colors = ('red', 'orange', 'green', 'blue', 'purple')

for iteration in range(10):
    color = colors[iteration % 5]
    
    plt.plot(g, 0, marker='o', color=color)
    y = f(g)
    plt.plot((g,g), (0,y), linestyle=':', color=color)
    plt.plot(g, y, marker='o', color=color)

    s = df(g)
    p = (g, y)
    
    change = newton_raphson_step(f, df, g)
    segment = np.linspace(g, g+change)
    plt.plot(segment, line(p, s, segment), linestyle='-', color=color)
    
    g += change
    
plt.grid()

plt.xlim((-5,5))
plt.ylim((-5,25))

# plt.xlim((-0.5,-0.4))
# plt.ylim((-0.25,0.25))
```

# The Economic Question

How many times do you have to apply the Newton-Raphson update algorithm before you're happy with your mistakes.

```python
g = 1.01

steps = 16

G = np.zeros(steps)
F = np.zeros(steps)

for iteration in range(steps):
    G[iteration] = g
    F[iteration] = f(g)
    
    change = newton_raphson_step(f, df, g)
    # print('   ',change)
    g += change

print(g, f(g))

correct_value = G[-1]
mistake = np.abs(G - correct_value) # absolute value
```

Log plots have axes where every unit of distance multiplies the value by some fixed amount (rather than add---a 'linear axis' adds).

```python
plt.plot(mistake)
plt.yscale('log')
plt.xlabel('Iteration')
plt.ylabel('Size of Mistake')
```

The rate at which our mistake shrinks is improving!
