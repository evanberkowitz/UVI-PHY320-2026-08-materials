# Good Code

In some sense, good code solves a problem.
Some people have the attitude that the problem-solving quality of a code is the *only* metric by which to judge code.
But there are other metrics by which we can judge code.

## Good code is *concise*.
Consider the difference between these two implementations of the same mathematical function,

```python
def isEven(n):
	if (n==1): return False
	if (n==2): return True
	if (n==3): return False
	if (n==4): return True
	if (n==5): return False
	if (n==6): return True
	# ...
	# the remainder of the code is omitted for brevity's sake

def isEven(n):
	return (n % 2) == 0
```

where the percent `%` symbol is the the *modulo* (or *remainder*) operator.
You can see, of course, that inevitably the first implementation will be deficient, because presumably the author only had a finite amount of time to `fill out the list' and must have ultimately missed some integers.

In fact, even if the first implementation were 'complete' it would be unsatisfactory, because it's nothing other than a big list of facts.
Don't get me wrong: sometimes you need to write some explicit facts down.
But *writing down the answer* or *looking up the answer* is not the same as *being able to figure out the answer*.
Good code is *instructive* and by studying someone else's code you might hope to learn something you didn't know before, even without running it.

## Good code is *clear*.

Consider the difference between these two implementations of the same mathematical function,

```python
def f(x):
	return 2*(x+1) - 2

def double(x):
	return 2*x
```

If you study them both it's clear that the `f` function and `double` function really *are* the same mathematical function.
But it's not obvious---for a few reasons.

First, the name `f` doesn't clearly express what the function *does*.
That's not a problem when you're looking at `f`'s definition, but it *could* be a problem when you use `f` somewhere else in your code far away from the definition!
In contrast if in that same spot you saw `double' the meaning would be obvious, assuming the person who named the function isn't a maniac.
Good code is *unsurprising* and *assigns meaningful names*.

Second, once you understand the use of `f` its *implementation* (the details of how it gets to its ultimate result) is extremely surprising!
Why go through all the rigamaroll of increase the argument by 1, doubling, and *then* subtracting 2?
The `double` function is also better on this count because its implementation is straightforward.

## Good code is *organized* and *well-documented*.

Code of course is written in specialized computer languages so that the computer can understand it as instructions.
But *humans* also read code.
In this course you should think of your audience as your classmate or instructor; in a programming job it might be your coworker or customer.
Writing code that is easy for these people to understand is important!
In this course it will help you earn a good grade; in the workplace it makes you a good team player.
Moreover, you'll be surprised by how easy it is to forget exactly how a function you yourself wrote works.

Most humans don't have the ability to keep a huge amount of detail in the front of mind at all times.
Instead most people can reason at a high level and drill down into the details of one thing at a time.
So it helps to keep related pieces of the code---where to understand one part you'd better understand the details of another---together.

For simple stuff like `isEven` or `double` the code is so simple that reading and understanding it is no issue.
But not all code is so short and sweet.
You are encouraged to use *comments* to explain non-obvious parts of your code.
Comments are parts of your code that the computer ignores when following the instructions in your code.
The lookup-table implmentation of `isEven` ends with two lines of comments, for example.
Comments should help make the idea explicit, the code more understandable, and should help clear up any possible confusions.

Good code *communicates*.

# Notebooks

This course's material will often be presented as a [jupyter](https://jupyter.org/) notebook---a sort of computational essay.
Notebooks are a form of [*literate programming*](https://en.wikipedia.org/wiki/Literate_programming)---the practice of interweaving a coherent plain-language *explanation* of code (and in the case of scientific programming, results) with the computer code itself.

Below I present what I would consider a good example of literate programming in a jupyter notebook.

Notice that it provides
 - Motivation,
 - Mathematics,
 - An explanation of the idea (separate from the explanation of the code),
 - Code with comments explaining the choices that went into it,
 - and even a demonstration of its correctness for a few examples.

Not all code needs to be presented in this way (for example, there might not *be* mathematics for the function that turns a string of characters into all UPPERCASE), but the presentation should be adapted to the audience, the purpose and use of the code, and whatever other considerations you can think of.

## Example: factorial

The *factorial* function arises throughout in combinatorics; it is equal to
\begin{equation}
    n! = n \times (n-1) \times (n-2) \times ... \times 1
\end{equation}
so that $1!=1$, $2!=2$, $3!=6$, $4!=24$ and so on.

A useful identity for computing $n!$ is that 
\begin{equation}
    n! = n \times (n-1)!
\end{equation}
which suggests a recursive implementation, where we build up the factorial step-by-step.  People even define 0!=1, which is compatible with the identity.

```python
def factorial(n):
    # We can't just use n! = n * (n-1)! forever---we need to know where to stop.
    # In particular we will need to encode *some* of the facts like 1! = 1, 2! = 2, 3! = 6, and 4! = 24.
    # But, the more facts we hard code the more likely it is we make an error.
    # So we'll just use the minimal required fact,
    if n == 0:
        return 1

    # And otherwise we'll rely on the recursive identity.
    return n * factorial(n-1)
```

To ensure that we've implemented `factorial` correctly let's check it against [A000142](https://oeis.org/A000142).

```python
A000142 = (1, 1, 2, 6, 24, 120, 720, 5040, 40320, 362880)

for n, answer in zip(range(10), A000142):
    f = factorial(n)
    if f == answer:
        print(f'factorial({n}) = {answer}')
    else:
        raise ValueError(f'Our factorial function disagrees with the known answer; factorial({n}) = {f} rather than {answer}')
```

# Libraries

Notebooks aren't always the perfect way to present code.  The issue is that the order you read or write them can be different from the order you *evaluate* them.  Consider this code snippet and think about what you should see printed before executing it.

```python
x=1
print(x)
x=2
```

Hopefully the answer was 'obvious' to you.  But if those same three lines were in different blocks

```python
x=1
```

```python
print(x)
```

```python
x=2
```

and then you were to *go back and re-evaluate the print statement* you might be surprised. by what you see.

The issue is that that notebooks have some *presentation* that does not necessarily match the *evaluation*.  This can make the results you see when you interact with a notebook dependent on your own personal behavior.

That's OK when you're actively developing an idea, but ultimately you'd like to land on code that reliably produces the same result every time, without human intervention.

Therefore, if you've developed your computation in a notebook and it finally works the right way, you should use it to construct a *library* which in python is called a *module*.  Maybe the particular computation you've done used specifics of the problem you're interested in but the code you've written can be reused for other problems.  That code should be put into the library; it might require a little bit of massaging so that all of the assumptions about your particular problem are 'factored out'.
