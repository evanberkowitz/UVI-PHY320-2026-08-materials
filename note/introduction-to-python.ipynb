The most important datatype for scientific programming is the *float*, which encodes [floating-point numbers](https://en.wikipedia.org/wiki/Floating-point_arithmetic), which are sort of like decimal numbers.

````python
float = (0., 0.1, 0.01, 0.01012301, 3.1415926535)
````

On most platforms floats are either 32 or 64 bits of data.  Those bits are used as follows

 - 1 bit for the sign
 - some bits for the 'significand'
 - some bits for the 'exponent'

and the number is expressed in base-2 scientific notation,
\begin{align}
    \text{float} = (\text{sign}) * (\text{significand}) * 2^{\text{exponent}}.
\end{align}

This is completely transparent to the user---you don't need to know the exact format of the floating point numbers, but you should be *aware* that this means there are, for example, a lot more floats between -1 and 1 than there are between 1023 and 1025 even though both of those intervals are length 2.

There are a lot of subtleties with floats, because humans like to write numbers in decimal but computers like to store data in binary.  One repercussion is that there are some real numbers that can't be exactly encoded as you might expect.

You can add 

````python
0.1+0.2
````

The 'mistake' arises from using a finite number of bits instead of *all* the binary digits.

The same mistake happens in base-10 if we 'chop off' the numbers after some number of digits.

For example, consider
```
 infinite   finite
 precision  precision
   
 1/3        0.3333333
+2/3       +0.6666666
====       ==========
 3/3 = 1   0.99999999, not 1
```
where we see that we can make a tiny mistake if we don't account for all the (infinite) digits after the decimal point.

The floating points numbers we use by default in python are 64 bits,

64 = (1 bit for the sign) + (52 bits for the significand) + (11 bits for the exponent)

The exponent can span from approximately $2^{10}$ to $-2^{10}$.  Since the exponent is itself in the exponent (!), that means $2^{\text{exponent}}$ can reach between $10^{\text{around }300}$ and $10^{\text{around }-300}$.

Despite the significand taking up 52 bits of memory, the significand is actually 53 bits.  What?  How?  The point is that in this 'binary scientific notation' the digit to the left of the decimal point is always a 1.  So we don't need to store that in memory, it'd be a waste.

Therefore, the amount of precision (where we 'chop off the numbers') in base 10 is

````python
significand_bits = 53
2**(-significand_bits)
````

In other words, the 'mistake' we expect when doing a single arithmetic operation on these numbers is 1 part in $10^{16}$.  It's a relative mistake---not an absolute mistake---in the sense that that the number of places (base 10) between the biggest nonzero digit and the mistake is 16 places.  (In base 2 it is 53 places!)

````python
(0.1+0.2)*100
````

## Assertions

An *assertion* is an instruction to check whether a value is true or false, and to crash the program if it is false and do nothing if it is true.

The general format is

```python
assert (predicate), "Optional error message"
# If you don't put the error message the comma isn't needed.
```

Assertions are useful when trying to hunt bugs.

````python
assert True
assert 1+1 == 2, "Uh-oh, math is broken!"
````

````python
assert False, "Crash the program on purpose!"
````

## Truthy and Falsey

Python has the boolean values of `True` and `False` and comparisons yield one of those two values.  But it's often convenient to *treat* other kinds of values as sort-of-representing-`True` and sort-of-representng-`False`.

We call these values [*Truthy* and *Falsey*](https://stackoverflow.com/questions/39983695/what-is-truthy-and-falsy-how-is-it-different-from-true-and-false) (It seems like the cases of these don't match---why is it not *Truey*?  Or *Falsehoody*?)

We can use `not` to flip `True` to `False` and vice-versa.

We can check truthiness with *assertions* which do nothing if the assertion is `True` and raise an error if the assertion is `False`.

`False` is obviously falsey while `True` obviously has truthiness.  Also `not` flips the truthiness.

````python
assert not False   # False is falsey;
assert True        # True has truthiness.
````

All integers have truthiness except 0 which is falsey.

````python
assert not 0       # 0 is falsey;
assert 12345       # any other integer has truthiness.
assert -67890
````

All strings have truthiness except the empty string `""` which is falsey.

````python
assert not ""      # The empty string is falsey;
assert "foo"       # any other string has truthiness.
assert "This is the 5nd week of class"
````

Other data types can also be interpreted as truthy and falsey.

````python
assert 1.0234
assert 0.1+0.2-0.3
````

## Variables

It's often beneficial to be able to repeat programs using the same instructions but different data.  So rather than type the data in, we give the data a *name*.  The *name* is the variable and it can refer to different data.

Then, you can use that data by just specifying its name, rather than its value.

````python
# So we can say
x = 123
# and then use the value that x refers to repeatedly
print(x)
print(x/3)
print(x+1)
print(x*2)
````

Some computer languages are really specific and picky about types.  Python is not.  It is *dynamically typed* which means that any variable can refer to any type.  It's perfectly valid to change the type of data a variable refers to.

````python
y = 123
print(y)
print(3*y)

y = "foo"
print(y)
print(3*y)
````

## Comparisons

We often need to make *comparisons* between different values.  The comparisons can yield a 'final answer', or help us determine what to do in pursuit of such an answer.

We can test for both equality and inequality, exactly as you might expect.

The result of a comparison is either `True` or `False`.

````python
print(1+1==2) # == is the 'check for equality' comparison
print(0==1)
assert (1 + 2 == 3)  # left == right --> is the left equal to the right
assert (1 + 2 < 4)   # left <  right --> is the left less than the right?
assert (1 + 2 <= 3)  # left <= right --> is the left less than or equal to the right?
assert (2 + 2 > 3)
assert (3 < 2 + 2)
assert (2 + 2 >= 4)
assert not (0 == 1)
````

but, sometimes you might be surprised when comparing floating-point numbers

````python
assert (0.1 + 0.2) == 0.3, "Floating point error"
````

String comparison uses [lexicographic order](https://en.wikipedia.org/wiki/Lexicographic_order),

````python
assert "Hello".upper() == "HELLO"
assert "a" < "b"
assert "ac" > "ab"
assert "foo" < "food"
````

## Control Structures

`If` and `else` let you execute different code based on some condition being true.

````python

today = 'Wednesday'

lots_of_business = {'Saturday', 'Sunday', 'Thursday', 'Friday'}

if today in lots_of_business:
    print("We're rolling in the dough!")
else:
    print("We're closed today.")

print("And have a nice day!")

# Notice: the whitespace matters!

# JAVA equivalent
# if (condition) {
#     what to do if the condition is true;
# }
# else {
#     what to do if the condition is false;
# }

````

````python
if today in lots_of_business:
    print("We're rolling in the dough!")
elif today == 'Wednesday':
    print("It's mid-week so we're open.")
else:
    print("We're closed today.")
````

A `for` loop is useful when you need to on a known collection of things.

````python
for day in lots_of_business:
    print(day.upper() + "!!!")
````
