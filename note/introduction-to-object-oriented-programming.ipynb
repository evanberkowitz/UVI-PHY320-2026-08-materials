Data is stored as long sequences of bits, but the computer also needs to know how to "make sense" of those bits---how to interpret themn, what kinds of manipulations are allowed, and so on.

In the code
```python
x = 4
print(type(x))
# <class 'int'>
```

There's a lot going on here:

 - x is the *name* of the variable, it labels some part of the computer's memory where the data we're talking about is stored.
 - 4 is the *value*---the "interpretation" of the bits.
 - `type(x)` asks the computer: hey, how are you "thinking about" this data.  The instructions for handling `x` are also stored in the computer's memory, and `type(x)` refers to that place in memory where we can look up those instructions.
  - `int` is the *class* of x---it's a *value* you find when you look up the type.

````python
import numpy as np

x = np.arange(10)
print(type(x))
````

````python
class Person:

    def __init__(self, id_number, name):
        self.id = id_number
        self.name = name

    def greet(self):
        # this is the default implementation of greet()
        print(f"Good day, {self.name}!")

author = Person(1234567890, "Evan Berkowitz")
author.greet()
````

The `author` is an *object* and that object is of the `Person` *class*.  The object is the *value*, the *class* is the idea---how to make sense of the data.

The class can provide instructions (functions) that can be called 'from' objects.  Since the object's type is the class, they know how to look up the functions in that class and apply those functions to themselves.  Hence the `me.greet()` pattern: `object.class_function(...)`.

# Inheritance

````python
class Student(Person):
    # what this means is that the Student class is a subclass of Person.
    # Every Student is a Person but not every Person is a Student.

    def __init__(self, id_number, name, year):

        self.year = year
        super().__init__(id_number, name)

    def greet(self):
        # students don't use the default greet()ing; they have a custom greet()ing.
        print("Good day,", self.name, "have a good", self.year, "year!")

    def years_to_graduation(self):
        # students also have an expected graduation time.
        if self.year == 'senior':
            return 0
        elif self.year == 'junior':
            return 1
        elif self.year == 'sophomore':
            return 2
        elif self.year == 'freshman':
            return 3
        else:
            raise ValueError(f'Unknown number of years to graduate from {self.year} year')
````

````python
you = Student(2345678901, "John Doe", 'senior')
classmate = Student(3456789012, "Joe Schmoe", 'junior')
puppet = Student(3456789012, "Count von Count", 'vampire')
````

````python
for person in (author, you, classmate, puppet):
    person.greet()
````

````python
for student in (you, classmate, puppet):
    print(student.name, "will graduate in", student.years_to_graduation(), "year's time!")
````

An *instance* is an object (value) that was created according to a particular *class*.  We can check whether the object *instantiates* the class (it's one of all the infinite (?) possibilities of the objects in that class) using the `isinstance` built-in python function.

````python
assert isinstance(you, Student) and isinstance(you, Person)
assert isinstance(author, Person)
assert isinstance(author, Student), "The author is not a Student."
````

````python
class Professor(Person):

    def campus(self, office):
        # Here's an example where the object's method uses data
        # - internal to the object (self.name) and
        # - external to the object (office).
        
        if 'RTP' in office:
            return self.name + ' sits on the AAS campus'
        if 'EVC' in office:
            return self.name + ' sits on the AAS campus'
        else:
            return self.name + ' sits on the OEK campus'
````

````python
eb = Professor(1234567890, 'Evan Berkowitz')
bg = Professor(5678901234, 'Bruce Gendre')

eb.campus('RTP 137')
bg.campus('CAB 105')
````
