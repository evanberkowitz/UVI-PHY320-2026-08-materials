Then commit it, **with a message that says what you did**:


Now change it — add a sentence, fix a typo, anything — and commit that separately:

```bash

```

Two commits, not one. The habit of committing each time a piece is finished, rather than
once at the very end, is worth more to you than any single git command. It gives you
somewhere to retreat to when the next change breaks everything.
