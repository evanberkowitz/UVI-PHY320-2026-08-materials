## Log Plots

Many scientific functions are exponentials.  They can get very big or very small without changing their input variable very much.  This makes them hard to plot clearly.

Let's imagine trying to plot
\begin{align}
    g(x) = e^{x + \sin(x)}
\end{align}

```python
def g(x):
    # You already know np.sin; the exponential e^x is written np.exp(x)
    return np.exp(x + np.sin(x))

x = np.linspace(-10, 10)
plt.plot(x, g(x))
```

We know the form of the function and we know there's some sine dependence in there, but from the above plot we don't see any evidence of the wavyness.  The problem is that the exponential growth swamps out the details.

We can make this easier to see by *rescaling* the y-axis.  Instead of plotting $y=g(x)$ we can instead plot $y =\log g(x)$.  The result is a [*log plot*](https://en.wikipedia.org/wiki/Logarithmic_scale) (as opposed to a linear plot).  Every step along the y-axis multiplies by the same factor (whereas we 'normally' add for each step).

You can take logarithm this by hand,
\begin{align}
    \log g(x) =x + \sin x
\end{align}

```python
plt.plot(x, x + np.sin(x))
```

but it's hard to interpret the y-axis.  Because it's not showing us the value of $g$ but the value of $\log(g)$.

We can instead just tell the library: "Hey, please rescale this function with the logarithm."

```python
plt.plot(x, g(x))
plt.yscale('log')
plt.grid()
```

Notice that it rescales log-base-10 instead of log-base-e (or ln).  But, because
\begin{align}
    \log_a x = \frac{\log_b x}{\log_b a}
\end{align}
this just scales the y-axis without changing the shape of the curve.

# Newton-Raphson

In class we used the Taylor series
\begin{align}
    f(x_0  + \Delta x) &\approx f(x_0) + f'(x_0) \Delta x + \frac{1}{2!} f''(x_0) \Delta x^2 + \cdots
\end{align}
to try and find a *zero* (or *root*) of a function $f$.

The idea is that we have a starting guess $x_0$ that we need to fix up by some amount $\Delta x$.
We are interested in the value of $\Delta x_0$ such that $f(x_0+\Delta x) = 0$---in other words $x_0+\Delta x$ is a root of $f$!

By truncating the Taylor series at first order (ie. ignoring the terms $\Delta x^2$ and higher), using $f(x_0 + \Delta x) = 0$ and solving for $\Delta x$ one finds
\begin{align}
    \Delta x_0 &= - \frac{f(x_0)}{f'(x_0)}
\end{align}

Suppose you won't be happy until you have in your hands an $x$ that produces $|f(x)| < \epsilon$ for some very small $\epsilon$.

Then:
1.  Start with any $x_0$.
2.  Evaluate $f(x_0)$.  If $|f(x_0)| < \epsilon$ celebrate; you are done.
3.  Evaluate the slope $f'(x_0)$.
4.  $\Delta x = -f(x_0) / f'(x_0)$
5.  Go back to step 1 but with an *updated* starting point: let $x_0$ be $x_0 + \Delta x$ instead.

```python
def newton_raphson(f, df, starting_guess, satisfying_smallness=1e-12):
    # starting_guess is our initial x_0
    # Let's check whether it's a zero (or close enough to satisfy our needs)
    height = f(starting_guess)

    if np.abs(height) < satisfying_smallness:
        return starting_guess

    slope = df(starting_guess)
    change = - height / slope
    return newton_raphson(f, df, starting_guess + change, satisfying_smallness)
```

The above implementation is *recursive*: it is defined in terms of itself!  We can always define recursive functions without recursion.  Which presentation / implementation is clearer depends on what the function is doing (and your own taste).  Here is a recursion-free implementation:

```python
def newton_raphson(f, df, starting_guess, satisfying_smallness=1e-12):
    # starting_guess is our initial x_0
    # Let's check whether it's a zero (or close enough to satisfy our needs)
    
    guess = starting_guess
    height = f(guess)
    while satisfying_smallness < np.abs(height):
        slope = df(guess)
        change = - height / slope
        guess += change
        height = f(guess)
    return guess
```

We checked in class that it could find zeros of a quadratic polynomial that has a real root.

But it works for other functions too.
